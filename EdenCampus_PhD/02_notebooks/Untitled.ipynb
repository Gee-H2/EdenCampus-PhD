```python
import os

# Check where the copy landed
github_repo = os.path.expanduser("~/Documents/YOUR_EXISTING_REPO_NAME")
destination = os.path.join(github_repo, "EdenCampus_PhD")

print("Checking destination folder...")
if os.path.exists(destination):
    files = []
    for root, dirs, filenames in os.walk(destination):
        for f in filenames:
            files.append(os.path.join(root, f))
    print(f"Found {len(files)} files at: {destination}")
else:
    print(f"Folder does NOT exist at: {destination}")
    print("The copy did not go to the right place.")

# Show all folders in Documents to find your repo
docs = os.path.expanduser("~/Documents")
print("\nFolders in your Documents:")
for item in sorted(os.listdir(docs)):
    full = os.path.join(docs, item)
    if os.path.isdir(full):
        print(f"  {item}")
```

```python
import os

gitignore_path = os.path.expanduser(
    "~/Documents/GitHub/EdenCampus-PhD/.gitignore")

# Read existing .gitignore if it exists
existing = ""
if os.path.exists(gitignore_path):
    with open(gitignore_path, 'r') as f:
        existing = f.read()

# Add .DS_Store if not already there
if ".DS_Store" not in existing:
    with open(gitignore_path, 'a') as f:
        f.write("\n# Mac system files\n.DS_Store\n.DS_Store?\n**/.DS_Store\n")
    print(".DS_Store added to .gitignore — will never appear again.")
else:
    print(".DS_Store already in .gitignore.")
```
